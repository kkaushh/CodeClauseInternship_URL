```python
import string
import random

# Dictionary to store URL mappings
url_database = {}

# Characters for short URL encoding (Base62)
characters = string.ascii_letters + string.digits
short_url_length = 6

def generate_short_code():
    """Generates a random short code of fixed length."""
    return ''.join(random.choices(characters, k=short_url_length))

def shorten_url(long_url):
    """Creates a short URL for a given long URL."""
    # Check if URL already exists
    for code, url in url_database.items():
        if url == long_url:
            return f"http://short.url/{code}"

    # Generate a new short code
    while True:
        short_code = generate_short_code()
        if short_code not in url_database:
            url_database[short_code] = long_url
            break

    return f"http://short.url/{short_code}"

def expand_url(short_url):
    """Expands a short URL back to its original long URL."""
    short_code = short_url.split('/')[-1]
    return url_database.get(short_code, "URL not found!")

# Example usage
if __name__ == "__main__":
    long_url = input("Enter the long URL: ")
    short = shorten_url(long_url)
    print("Shortened URL:", short)

    print("\nNow let's try to expand it...")
    expanded = expand_url(short)
    print("Expanded URL:", expanded)
```
